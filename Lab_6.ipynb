```python
from langchain_classic.memory import ConversationSummaryMemory
from langchain_classic.chains import ConversationChain
```

```python
GROQ_API_KEY = "AQ.Ab8RN6LrjJlXZ4pxTXpcz8n9sReC............................"

grok_model = ChatGroq( model="llama-3.3-70b-versatile",
                      api_key=GROQ_API_KEY,
                       )
```

```python
memory=ConversationSummaryMemory(
    memory_key="history",
    return_messages=True,
    llm=grok_model)
```

```python
pipeline = ConversationChain(
    llm=grok_model,
    memory=memory,
    verbose=True
)

while True:
  user_input=input("User:")
  if user_input=='exit':
    break
  response=pipeline.invoke(input=user_input)
  print(response)
```

### Securely Storing API Keys with `google.colab.userdata`

1.  **Add your API Key to Colab Secrets:**
    *   On the left sidebar, click the "🔑 Secrets" icon.
    *   Click "+ New Secret".
    *   For the "Name" field, enter a name for your API key (e.g., `GROQ_API_KEY`).
    *   For the "Value" field, paste your actual API key.
    *   Make sure to toggle "Notebook access" to ON for the current notebook.

2.  **Access the API Key in your notebook:**
    Once stored, you can access your API key in your code using `userdata.get('YOUR_SECRET_NAME')`.

```python
# Import the userdata module
from google.colab import userdata

# Retrieve your stored API key
# Replace 'GROQ_API_KEY' with the actual name you gave your secret
GROQ_API_KEY_SECURE = userdata.get('GROQ_API_KEY')

# You can now use GROQ_API_KEY_SECURE in your `ChatGroq` constructor
print("API Key successfully retrieved from Colab secrets.")
# For security, avoid printing the key directly in your output.
```
