```python
from langchain_classic.memory import ConversationBufferWindowMemory
from langchain_classic.chains import ConversationChain
```

```python
GROQ_API_KEY = "AQ.Ab8RN6LrjJlXZ4pxTXpcz8n9sReC.........................."

grok_model = ChatGroq( model="llama-3.3-70b-versatile",
                      api_key=GROQ_API_KEY,
                       )
```

```python
memory=ConversationBufferWindowMemory(
    memory_key="history",
    return_messages=True,
    k=2
)
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["history", "input"],
    template=system_prompt_text + "\n\n{history}\nHuman: {input}\nAI:"
)

pipeline=ConversationChain(
    llm=grok_model,
    memory=memory,
    prompt=prompt,
    verbose=True
)
```

```python
while True:
  user_input=input("USER:")
  if user_input=='exit':
    break
  response = pipeline.invoke(input=user_input)

  print("AI:",response)
```

```python
system_prompt_text = """The following is a friendly conversation between a human and an AI. The AI is talkative and provides lots of specific details from its context. If the AI does not know the answer to a question, it truthfully says it does not know."""
```
